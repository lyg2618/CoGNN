```python
import os
os.environ["CUDA_VISIBLE_DEVICES"] = '0'
```

```python
import torch
import numpy as np
from torch_geometric.data import Data

from utils import Dataset, PMI_score, get_edge_index, setup_seed
from model import CoGNN
from train import train
```

```python
dataset_path = 'data/100k'

# Load the dataset
dataset = Dataset(dataset_path)
train_mat, test_ratings, test_negatives = dataset.trainMatrix, dataset.testRatings, dataset.testNegatives
print('Dataset: #user=%d, #item=%d, #train_pairs=%d, #test_pairs=%d' 
      % (dataset.num_users, dataset.num_items, train_mat.nnz, len(test_ratings)))

mat = train_mat.todense()
mat = 1 * (mat > 0)
if type(mat) is np.matrix: mat = np.asarray(mat)
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

users_num, items_num = dataset.num_users, dataset.num_items
```

- k = 1
- is_normalised=False

```python
k = 1

NPMI_u, NPMI_i = PMI_score(mat, k=k, is_normalised=False)

edge_index_u, edge_index_i = get_edge_index(NPMI_u), get_edge_index(NPMI_i)
x_u, x_i = torch.tensor(mat, dtype=torch.float), torch.tensor(mat.T, dtype=torch.float)
data_u, data_i = Data(x=x_u, edge_index=edge_index_u, edge_attr=NPMI_u), Data(x=x_i, edge_index=edge_index_i, edge_attr=NPMI_i)
```

```python
lr = 0.001
alpha = 0.01
weight_decay = 1e-6
embedding_size_ae = 256
embedding_size_graph = 32
hidden_dim = 64
num_layers_mlp = 1
num_layers_transformer = 1
epochs = 32
batch_size = 1024
mode = 'hr'
topK = 10
Loss = 'BCE'

hr_list, ndcg_list = [], []
for seed in range(3):
    setup_seed(seed)
    model = CoGNN(rating_mat=torch.from_numpy(mat).float().to(device), 
                  data_u=data_u.to(device), 
                  data_i=data_i.to(device), 
                  embedding_size_ae=embedding_size_ae, 
                  embedding_size_graph=embedding_size_graph, 
                  hidden_dim=hidden_dim,
                  num_layers_transformer=num_layers_transformer, 
                  num_layers_mlp=num_layers_mlp)
    model = model.to(device)
    hr, ndcg = train(model=model, 
                      lr=lr, 
                      weight_decay=weight_decay, 
                      test_ratings=test_ratings, 
                      test_negatives=test_negatives, 
                      topK=topK, 
                      mat=mat, 
                      epochs=epochs, 
                      alpha=alpha,
                      batch_size=batch_size,
                      mode=mode, 
                      device=device,
                      Loss=Loss)
    hr_list.append(hr)
    ndcg_list.append(ndcg)

print('alpha: ', alpha, 'embedding_size_graph: ',embedding_size_graph, 'num_layers_mlp: ', num_layers_mlp)
print('HR: %.4f' %np.mean(hr_list), 'NDCG: %.4f' %np.mean(ndcg_list))
print('=' * 100)
```
